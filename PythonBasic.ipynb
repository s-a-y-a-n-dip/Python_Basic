```python
# 10. Write code to demonstrate type casting with list elements.
# Ans. 

string_list = ["10", "20", "30", "40", "50"]
int_list = [int(x) for x in string_list]
print("List after converting to integers:", int_list)
float_list = [float(x) for x in string_list]
print("List after converting to floats:", float_list)
mixed_list = ["5", 10.5, "15", 20.0, "25"]
int_mixed_list = [int(float(x)) for x in mixed_list]
print("Mixed list after converting to integers:", int_mixed_list)
float_mixed_list = [float(x) for x in mixed_list]
print("Mixed list after converting to floats:", float_mixed_list)
bool_string_list = ["True", "False", "1", "0", "yes", "no"]
def to_boolean(value):
    if value.lower() in ("true", "1", "yes"):
        return True
    elif value.lower() in ("false", "0", "no"):
        return False
    else:
        return None

bool_list = [to_boolean(x) for x in bool_string_list]
print("List after converting to booleans:", bool_list)
```

```python
# 14. Write a program to reverse a string using a while loop.
# Ans. 

def reverse_string(s):
    reversed_str = ""
    index = len(s) - 1
    while index >= 0:
        reversed_str += s[index]
        index -= 1
    return reversed_str
string = input("Enter a string: ")
print("Reversed string:", reverse_string(string))
```

```python
# 15. Write a Python program to calculate the factorial of a number provided by the user using a while loop.
# Ans. 

def factorial(n):
    if n < 0:
        return "Factorial is not defined for negative numbers."  
    result = 1
    while n > 0:
        result *= n
        n -= 1
    return result
num = int(input("Enter a number: "))
print(f"The factorial of {num} is: {factorial(num)}")
```
